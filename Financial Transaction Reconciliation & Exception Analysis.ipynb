Ledger vs bank amount inconsistencies.

Identify Records with Missing Amounts

```python
missing_amount_records = df[
    df["LEDGER_AMOUNT"].isna() |
    df["BANK_AMOUNT"].isna()
]
missing_amount_records.shape
```

Identify Amount Mismatches

```python
AMOUNT_TOLERANCE = 0.01
currency = (
    df["CURRENCY"]
    .astype("string")
    .str.strip()
    .str.upper()
)
bank_currency = (
    df["BANK_CURRENCY"]
    .astype("string")
    .str.strip()
    .str.upper()
)
same_currency = (
    currency.notna()
    & bank_currency.notna()
    & currency.ne("")
    & bank_currency.ne("")
    & currency.ne("UNKNOWN")
    & bank_currency.ne("UNKNOWN")
    & currency.eq(bank_currency)
)
comparable_amounts = df[
    df["LEDGER_AMOUNT"].notna()
    & df["BANK_AMOUNT"].notna()
    & same_currency
].copy()
```

Calculate the Difference

```python
comparable_amounts["AMOUNT_DIFFERENCE"] = (
    comparable_amounts["LEDGER_AMOUNT"] -
    comparable_amounts["BANK_AMOUNT"]
)
comparable_amounts["ABS_AMOUNT_DIFFERENCE"] = (
    comparable_amounts["AMOUNT_DIFFERENCE"].abs()
)
comparable_amounts.head(10)
```

Identify non zero Discrepancies

```python
amount_discrepancies = comparable_amounts[
    comparable_amounts["ABS_AMOUNT_DIFFERENCE"] > AMOUNT_TOLERANCE
].copy()
print("Amount tolerance:", AMOUNT_TOLERANCE)
print("Total amount discrepancies:", amount_discrepancies.shape[0])
display(amount_discrepancies.head(10))
```

```python
amount_discrepancies[
    [
        "TRANSACTION_ID",
        "TRANSACTION_TYPE",
        "TRANSACTION_STATUS",
        "CURRENCY",
        "LEDGER_AMOUNT",
        "BANK_AMOUNT",
        "AMOUNT_DIFFERENCE",
        "RECONCILIATION_STATUS"
    ]
]
```

Identify which have the higher amount

```python
print(
    "Ledger amount higher:",
    (amount_discrepancies["AMOUNT_DIFFERENCE"] > 0).sum()
)
print(
    "Bank amount higher:",
    (amount_discrepancies["AMOUNT_DIFFERENCE"] < 0).sum()
)
print(
    "Amount differences:",
    (amount_discrepancies["AMOUNT_DIFFERENCE"] == 0).sum()
)
```

Creating a final report

```python

transaction_dates = pd.to_datetime(
    df["TRANSACTION_DATE"],
    format="%d-%m-%Y",
    errors="coerce"
)

settlement_dates = pd.to_datetime(
    df["SETTLEMENT_DATE"],
    format="%d-%m-%Y",
    errors="coerce"
)

invalid_transaction_dates = (
    df["TRANSACTION_DATE"].notna()
    & transaction_dates.isna()
)

invalid_settlement_dates = (
    df["SETTLEMENT_DATE"].notna()
    & settlement_dates.isna()
)

settlement_before_transaction = (
    transaction_dates.notna()
    & settlement_dates.notna()
    & (settlement_dates < transaction_dates)
)

valid_status = ["Settled", "Pending", "Failed", "Reversed"]

normalized_status = (
    df["TRANSACTION_STATUS"]
    .astype("string")
    .str.strip()
    .str.title()
)

invalid_status = (
    normalized_status.isna()
    | ~normalized_status.isin(valid_status)
)

currency = (
    df["CURRENCY"]
    .astype("string")
    .str.strip()
    .str.upper()
)

bank_currency = (
    df["BANK_CURRENCY"]
    .astype("string")
    .str.strip()
    .str.upper()
)

currency_missing_or_unknown = (
    currency.isna()
    | bank_currency.isna()
    | currency.eq("")
    | bank_currency.eq("")
    | currency.eq("UNKNOWN")
    | bank_currency.eq("UNKNOWN")
)

currency_mismatch_mask = (
    ~currency_missing_or_unknown
    & currency.ne(bank_currency)
)

same_currency_mask = (
    ~currency_missing_or_unknown
    & currency.eq(bank_currency)
)

comparable_amounts = df[
    df["LEDGER_AMOUNT"].notna()
    & df["BANK_AMOUNT"].notna()
    & same_currency_mask
].copy()

comparable_amounts["AMOUNT_DIFFERENCE"] = (
    comparable_amounts["LEDGER_AMOUNT"]
    - comparable_amounts["BANK_AMOUNT"]
)

amount_discrepancies = comparable_amounts[
    comparable_amounts["AMOUNT_DIFFERENCE"].abs() > AMOUNT_TOLERANCE
].copy()

text_columns = df.select_dtypes(include="object").columns

extra_space_counts = {}

for col in text_columns:
    values = df[col].astype("string")
    extra_space_counts[col] = (
        values.notna()
        & values.ne(values.str.strip())
    ).sum()

total_formatting_issues = sum(extra_space_counts.values())

data_quality_summary = pd.DataFrame({
    "Assessment": [
        "Missing Values",
        "Duplicate Full Records",
        "Duplicate Transaction IDs",
        "Duplicate Reference IDs",
        "Invalid Transaction Dates",
        "Invalid Settlement Dates",
        "Settlement Before Transaction",
        "Negative Ledger Amounts",
        "Negative Bank Amounts",
        "Zero Ledger Amounts",
        "Zero Bank Amounts",
        "Invalid/Missing Transaction Status",
        "Currency Mismatch",
        "Same-Currency Amount Discrepancies",
        "Text Formatting Issues"
    ],
    "Finding": [
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["TRANSACTION_ID"].duplicated().sum(),
        df["REFERENCE_ID"].duplicated().sum(),
        invalid_transaction_dates.sum(),
        invalid_settlement_dates.sum(),
        settlement_before_transaction.sum(),
        (df["LEDGER_AMOUNT"] < 0).sum(),
        (df["BANK_AMOUNT"] < 0).sum(),
        (df["LEDGER_AMOUNT"] == 0).sum(),
        (df["BANK_AMOUNT"] == 0).sum(),
        invalid_status.sum(),
        currency_mismatch_mask.sum(),
        len(amount_discrepancies),
        total_formatting_issues
    ]
})

display(data_quality_summary)
```

Identify currency mismatches

```python
recon_df["CURRENCY_MISMATCH"] = (
    recon_df["CURRENCY"].notna()
    & recon_df["BANK_CURRENCY"].notna()
    & recon_df["CURRENCY"].ne("UNKNOWN")
    & recon_df["BANK_CURRENCY"].ne("UNKNOWN")
    & recon_df["CURRENCY"].ne(recon_df["BANK_CURRENCY"])
)
print("Currency mismatches:",
      recon_df["CURRENCY_MISMATCH"].sum())
```

Compare ledger and bank amounts

```python
same_currency = (
    recon_df["CURRENCY"].notna()
    & recon_df["BANK_CURRENCY"].notna()
    & recon_df["CURRENCY"].ne("")
    & recon_df["BANK_CURRENCY"].ne("")
    & recon_df["CURRENCY"].ne("UNKNOWN")
    & recon_df["BANK_CURRENCY"].ne("UNKNOWN")
    & recon_df["CURRENCY"].eq(recon_df["BANK_CURRENCY"])
)
comparable_mask = (
    same_currency
    & recon_df["LEDGER_AMOUNT"].notna()
    & recon_df["BANK_AMOUNT"].notna()
)
recon_df["AMOUNT_DIFFERENCE"] = np.nan
recon_df.loc[comparable_mask, "AMOUNT_DIFFERENCE"] = (
    recon_df.loc[comparable_mask, "LEDGER_AMOUNT"]
    - recon_df.loc[comparable_mask, "BANK_AMOUNT"]
)
recon_df["AMOUNT_MISMATCH"] = (
    recon_df["AMOUNT_DIFFERENCE"].abs() > AMOUNT_TOLERANCE
)
print("Amount tolerance:", AMOUNT_TOLERANCE)
print("Comparable transactions:", comparable_mask.sum())
print(
    "Amount mismatches:",
    recon_df.loc[comparable_mask, "AMOUNT_MISMATCH"].sum()
)
```

Exception Analysis

Categorize exceptions

```python
exception_df = recon_df[
    recon_df["RECONCILIATION_STATUS"] != "MATCHED"
].copy()
exception_summary = (
    exception_df.groupby("RECONCILIATION_STATUS")
    .agg(
        TOTAL_EXCEPTIONS=("TRANSACTION_ID", "count")
    )
    .reset_index()
    .sort_values("TOTAL_EXCEPTIONS", ascending=False)
)
exception_summary["EXCEPTION_PERCENTAGE"] = (
    exception_summary["TOTAL_EXCEPTIONS"]
    / len(recon_df) * 100
).round(2)
display(exception_summary)
```

Analyze exception severity

```python
severity_summary = (
    exception_df.groupby("EXCEPTION_SEVERITY")
    .agg(
        TOTAL_EXCEPTIONS=("TRANSACTION_ID", "count")
    )
    .reset_index()
)
severity_summary["PERCENTAGE"] = (
    severity_summary["TOTAL_EXCEPTIONS"]
    / len(exception_df) * 100
).round(2)
display(severity_summary)
```

Exception reason vs severity

```python
reason_severity = pd.crosstab(
    exception_df["EXCEPTION_REASON"],
    exception_df["EXCEPTION_SEVERITY"],
    margins=True
)
display(reason_severity)
```

Financial impact of amount mismatches

```python
financial_impact = (
    recon_df[recon_df["AMOUNT_MISMATCH"]]
    .groupby("CURRENCY")
    .agg(
        TOTAL_MISMATCHED_TRANSACTIONS=("TRANSACTION_ID", "count"),
        TOTAL_ABSOLUTE_DIFFERENCE=("AMOUNT_DIFFERENCE",
                                   lambda x: x.abs().sum()),
        AVERAGE_ABSOLUTE_DIFFERENCE=("AMOUNT_DIFFERENCE",
                                     lambda x: x.abs().mean())
    )
    .round(2)
    .reset_index()
)
display(financial_impact)
```

```python

def automate_rca(data):

    print("ROOT CAUSE ANALYSIS REPORT")
    print("=" * 40)

    print("\n1. Exception Reason Analysis")
    reason = data["EXCEPTION_REASON"].value_counts()
    print(reason)

    print("\n2. Severity Analysis")
    severity = data["EXCEPTION_SEVERITY"].value_counts()
    print(severity)

    print("\n3. Source System Analysis")
    source = data["SOURCE_SYSTEM"].value_counts()
    print(source)

    print("\n4. Transaction Type Analysis")
    transaction = data["TRANSACTION_TYPE"].value_counts()
    print(transaction)

    print("\n5. Currency Analysis")
    currency = (
        data["CURRENCY"]
        .astype("string")
        .str.strip()
        .str.upper()
        .replace({"EURO": "EUR"})
        .value_counts()
    )
    print(currency)

    print("\n6. Bank Analysis")
    bank = data["BANK_NAME"].value_counts()
    print(bank)

    print("\n7. Country Analysis")
    country = (
        data["COUNTRY"]
        .astype("string")
        .str.strip()
        .str.title()
        .replace({"Uk": "United Kingdom"})
        .value_counts()
    )
    print(country)

    print("\n8. Batch Analysis")
    batch = data["BATCH_ID"].value_counts()
    print(batch.head(10))

    return {
        "EXCEPTION_REASON": reason,
        "SEVERITY": severity,
        "SOURCE_SYSTEM": source,
        "TRANSACTION_TYPE": transaction,
        "CURRENCY": currency,
        "BANK": bank,
        "COUNTRY": country,
        "BATCH": batch
    }
```

```python
def automate_business_impact(df, exception_report):

    print("BUSINESS IMPACT ANALYSIS")
    print("=" * 40)

    total_transactions = len(df)
    total_exceptions = len(exception_report)

    exception_rate = (
        total_exceptions / total_transactions * 100
        if total_transactions > 0 else 0
    )
    currency = (
        exception_report["CURRENCY"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    bank_currency = (
        exception_report["BANK_CURRENCY"]
        .astype("string")
        .str.strip()
        .str.upper() 
    )
    
    comparable = (
        exception_report["LEDGER_AMOUNT"].notna()
        & exception_report["BANK_AMOUNT"].notna()
        & currency.notna()
        & bank_currency.notna()
        & currency.ne("UNKNOWN")
        & bank_currency.ne("UNKNOWN")
        & currency.eq(bank_currency)
    )

    amount_difference = (
        exception_report["LEDGER_AMOUNT"]
        - exception_report["BANK_AMOUNT"]
    ).where(comparable)

    amount_tolerance = 0.01

    amount_discrepancies = amount_difference[
        amount_difference.abs() > amount_tolerance
    ]

    total_discrepancy = amount_discrepancies.abs().sum()

    reason = (
        exception_report["EXCEPTION_REASON"]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    missing_bank = exception_report[
        reason == "BANK RECORD NOT FOUND"
    ]

    missing_bank_amount = (
        missing_bank["LEDGER_AMOUNT"].abs().sum()
    )
    
    missing_ledger = exception_report[
        reason == "LEDGER RECORD NOT FOUND"
    ]

    missing_ledger_amount = (
        missing_ledger["BANK_AMOUNT"].abs().sum()
    )

    severity = (
        exception_report["EXCEPTION_SEVERITY"]
        .value_counts()
    ) 
    print("Total Transactions:", total_transactions)
    print("Total Exceptions:", total_exceptions)
    print("Exception Rate:", round(exception_rate, 2), "%")

    print("\nAmount Discrepancies")
    print("Affected Transactions:", len(amount_discrepancies))
    print("Total Absolute Difference:", round(total_discrepancy, 2))

    print("\nMissing Bank Transactions")
    print("Transaction Count:", len(missing_bank))
    print("Ledger Amount:", round(missing_bank_amount, 2))

    print("\nMissing Ledger Transactions")
    print("Transaction Count:", len(missing_ledger))
    print("Bank Amount:", round(missing_ledger_amount, 2))

    print("\nException Severity")
    print(severity)

    return {
        "TOTAL_TRANSACTIONS": total_transactions,
        "TOTAL_EXCEPTIONS": total_exceptions,
        "EXCEPTION_RATE": round(exception_rate, 2),
        "AMOUNT_DISCREPANCIES": len(amount_discrepancies),
        "TOTAL_ABSOLUTE_DISCREPANCY": round(total_discrepancy, 2),
        "MISSING_BANK_COUNT": len(missing_bank),
        "MISSING_BANK_AMOUNT": round(missing_bank_amount, 2),
        "MISSING_LEDGER_COUNT": len(missing_ledger),
        "MISSING_LEDGER_AMOUNT": round(missing_ledger_amount, 2),
        "SEVERITY": severity
    }
```

```python
business_impact = automate_business_impact(
    df,
    exception_report
)
```

```python
from pathlib import Path
import pandas as pd

def generate_excel_report():

    report_path = Path(
        "Financial_Reconciliation_Project/Output/Reports"
    )

    report_path.mkdir(parents=True, exist_ok=True)

    file_path = report_path / "Financial_Reconciliation_Report_Final.xlsx"

    with pd.ExcelWriter(
        file_path,
        engine="openpyxl",
        mode="w"
    ) as writer:
        quality_report.to_excel(
            writer,
            sheet_name="Data Quality",
            index=False
        )
        exception_report.to_excel(
            writer,
            sheet_name="Exception Report",
            index=False
        )
        rca_sheet_names = {
            "EXCEPTION_REASON": "RCA - Reasons",
            "SEVERITY": "RCA - Severity",
            "SOURCE_SYSTEM": "RCA - Source",
            "TRANSACTION_TYPE": "RCA - Transactions",
            "CURRENCY": "RCA - Currency",
            "BANK": "RCA - Banks",
            "COUNTRY": "RCA - Countries",
            "BATCH": "RCA - Batches"
        }

        for key, sheet_name in rca_sheet_names.items():

            rca_df = (
                rca_results[key]
                .rename_axis(key)
                .reset_index(name="COUNT")
            )

            rca_df.to_excel(
                writer,
                sheet_name=sheet_name,
                index=False
            )

        impact_summary = {
            key: value
            for key, value in business_impact.items()
            if key != "SEVERITY"
        }

        impact_df = pd.DataFrame(
            list(impact_summary.items()),
            columns=["METRIC", "VALUE"]
        )

        impact_df.to_excel(
            writer,
            sheet_name="Business Impact",
            index=False
        )

        severity_df = (
            business_impact["SEVERITY"]
            .rename_axis("SEVERITY")
            .reset_index(name="EXCEPTION_COUNT")
        )

        severity_df.to_excel(
            writer,
            sheet_name="Impact Severity",
            index=False
        )

    print("Excel report generated successfully!")
    print("File location:", file_path)

    return str(file_path)
```

```python
quality_report = data_quality_report(df)
rca_results = automate_rca(exception_report)
business_impact = automate_business_impact(df, exception_report)

print("Excel export inputs are ready:")
print("quality_report:", quality_report.shape)
print("exception_report:", exception_report.shape)
print("RCA sections:", list(rca_results.keys()))
print("business_impact metrics:", list(business_impact.keys()))
```

```python
excel_file = generate_excel_report()
```
